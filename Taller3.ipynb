### Ejercicio1

```python
def newton_raphson(f, df, p0, TOL=1e-6, N0=50):
    """Algoritmo 2.3. Devuelve (p, historia), con historia = [p0, p1, ..., p]."""
    historia = [p0]
    i = 1
    while i <= N0:
        d = df(p0)
        if d == 0:
            raise ZeroDivisionError(f"f'(p0) = 0 en p0 = {p0}: la tangente es horizontal")
        p = p0 - f(p0)/d                              # COMPLETE: Paso 3, usando d
        historia.append(p)                            # COMPLETE
        if abs(p - p0) < TOL:                         # COMPLETE: Paso 4
            return p, historia
        i = i + 1
        p0 = p                                        # COMPLETE: Paso 6
    raise RuntimeError(f"El método fracasó después de {N0} iteraciones")

def f(x):
    return x**3 + 4 * x ** 2 - 10

def df(x):
    return 3 * x ** 2 + 8 * x

newton_raphson(f, df, 1.5)
```

### Ejecicio 2

```python

def secante(f, p0, p1, TOL=1e-6, N0=50):
    """Algoritmo 2.4. Devuelve (p, historia)."""

    # Paso 1
    i = 2
    q0 = f(p0)
    q1 = f(p1)
    historia = [p0, p1]

    # Paso 2
    while i <= N0:

        # Verificar división entre cero
        if q1 - q0 == 0:
            raise ZeroDivisionError(
                f"f(p1) - f(p0) = 0 en p0 = {p0}, p1 = {p1}: la secante es horizontal"
            )

        # Paso 3: calcular nueva aproximación
        p = p1 - q1 * (p1 - p0) / (q1 - q0)
        historia.append(p)

        # Paso 4: comprobar tolerancia
        if abs(p - p1) < TOL:
            return p, historia

        # Paso 5: incrementar contador
        i = i + 1

        # Paso 6: actualizar puntos
        p0, q0, p1, q1 = p1, q1, p, f(p)

    # Paso 7
    raise RuntimeError(
        f"El método fracasó después de {N0} iteraciones"
    )

llamadas = []

def f(x):
    llamadas.append(x)
    return x**3 + 4*x**2 - 10

llamadas.clear()         # vacíe la lista antes de cada prueba
p, historia = secante(f, 1.0, 2.0, TOL=1e-6)
print(p, historia, len(llamadas))

```

```python

# Función del saldo
def saldo_final(i, monto=20000.0, cuota=450.0, meses=60):
    saldo = monto
    for mes in range(meses):
        saldo = saldo * (1 + i) - cuota
    return saldo


# Método de la secante
def secante(f, p0, p1, TOL=1e-6, N0=50):
    i = 2
    q0 = f(p0)
    q1 = f(p1)
    historia = [p0, p1]

    while i <= N0:
        if q1 - q0 == 0:
            raise ZeroDivisionError("La secante es horizontal")

        p = p1 - q1 * (p1 - p0) / (q1 - q0)
        historia.append(p)

        if abs(p - p1) < TOL:
            return p, historia

        i = i + 1
        p0, q0, p1, q1 = p1, q1, p, f(p)

    raise RuntimeError("El método fracasó")


# Calcular la tasa
tasa, historia = secante(saldo_final, 0.005, 0.02, TOL=1e-10)

print("Tasa mensual (%):", tasa * 100)
print("Saldo final:", saldo_final(tasa))

# Tasa efectiva anual
tasa_anual = (1 + tasa)**12 - 1
print("Tasa efectiva anual (%):", tasa_anual * 100)
```
